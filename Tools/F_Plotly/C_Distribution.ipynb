## Distribution plots
- Histogram
- Box plot
- Violin plot
- KDE / Density plot
- ECDF plot

```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
student = pd.read_csv('datasets/student_dataset_complete.csv') 
student.head()
```

```python
px.histogram(student, x="Marks", title="Marks Distribution").show()
```

```python
px.box(student, y="Marks", title="Marks Spread").show()
```

```python
# KDE/Density with violin
px.violin(student, y="test_score", points=False, title="Density of Test Score").show()
```

```python
px.ecdf(student, x="Marks", title="ECDF of Marks").show()
```
