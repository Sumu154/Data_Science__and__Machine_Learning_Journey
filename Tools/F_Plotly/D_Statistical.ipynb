## Statistical plots
- Box plot
- Violin plot
- Histogram
- Error bar plot

```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
student = pd.read_csv('datasets/student_dataset_complete.csv') 
student.head()
```

```python
px.box(student, x="subject", y="Marks", title="Spread by Subject").show()
```

```python
px.violin(student, x="subject", y="Marks", box=True, title="Distribution by Subject").show()
```

```python
px.histogram(student, x="Marks", title="Marks Histogram").show()
```
