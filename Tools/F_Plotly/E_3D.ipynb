## Statistical plots
- 3D Scatter
- 3D Line
- 3D Surface
- 3D Mesh
- 3D Contour

```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
student = pd.read_csv('datasets/student_dataset_complete.csv') 
student.head()
```

```python
# 3D Scatter
px.scatter_3d(student, x="study_hours", y="Marks", z="attendance_rate", color="gender", title="3D Scatter Plot").show()
```

```python
# 3D Line
px.line_3d(student, x="week", y="Marks", z="study_hours", title="3D Marks Trend").show()
```

```python
# 3D Mesh
px.scatter_3d(student, x="week", y="Marks", z="study_hours").show()
```

```python
# 3D Contour
px.density_contour(student, x="study_hours", y="Marks", title="3D Contour Feel").show()
```
