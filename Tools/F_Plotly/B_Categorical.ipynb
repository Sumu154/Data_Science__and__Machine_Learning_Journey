## Categorical plots

- Bar chart
- Grouped bar
- Stacked bar
- Pie chart
- Donut chart
- Sunburst chart
- Treemap

```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
student = pd.read_csv('datasets/student_dataset_complete.csv') 
student.head()
```

```python
px.bar(student, x="subject", y="Marks", title="Average Marks by Subject").show()
```

```python
px.bar(student, x="gender", y="Marks", color="subject", barmode="group", title="Marks by Gender and Subject").show()
```

```python
px.bar(student, x="class_name", y="Marks", color="gender", barmode="stack", title="Marks by Class and Gender").show()
```

```python
px.pie(student, names="gender", title="Gender Distribution").show()
```

```python
px.pie(student, names="class_name", hole=0.5, title="Class Distribution").show()
```

```python
px.sunburst(student, path=["subject","class_name"], values="Marks",
            title="Marks by Subject & Class").show()
```

```python
px.treemap(student, path=["subject","gender"], values="Marks", title="Treemap of Marks by Subject-Gender").show()
```
