```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly as px
```

### 1. Relational Plot

- to see the statistical relation between 2 or more variables.
- Bivariate Analysis

Plots under this section

- scatterplot
- lineplot

```python
# sns datasets 
tips = sns.load_dataset('tips')
tips
```

```python
# 1) scatter plot 
# axes level function
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='time', size='size')
# figure level function -> relplot 
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex', style='time', size='size')
```

```python
# 2) line plot
gap = px.data.gapminder()
temp_df = gap[gap['country'].isin(['India', 'Brazil', 'Germany'])]
temp_df.head(5)
```

```python
# axes level function
sns.lineplot(data=temp_df, x='year', y='lifeExp', hue='country', style='continent', size='continent')
# figure level function
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line', hue='country', style='continent', size='continent')
```

```python
# *****facet plot***** -> figure level figure -> work with relplot
# it will not work with .scatterplot and .lineplot
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='smoker', row='sex')

# col wrap with facet
sns.relplot(data=gap, x='lifeExp', y='gdpPercap', kind='scatter', col='year', col_wrap=3)
```
