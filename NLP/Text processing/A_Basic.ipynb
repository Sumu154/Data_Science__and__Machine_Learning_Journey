```python
import numpy as np
import pandas as pd

import re
```

```python
df = pd.read_csv('datasets/IMDB Dataset.csv')
df.head(2)
```

```python
# Lowercasing
text = "I Love NLP!"
text.lower()  # Output: 'i love nlp!'

df['review'].str.lower()
# df['review'][3].lower()
```

```python
# 2. removing HTML tags
def remove_html_tags(text):
  pattern = re.compile(r'<.*?>')
  return pattern.sub(r'', text)

text = "<p>I love <b>NLP</b> and <i>text processing</i>.</p>"
cleaned_text = remove_html_tags(text)
# print(cleaned_text)

# function ta pass korte hbe
df['review'] = df['review'].apply(remove_html_tags)
df['review']
```

```python
# 3. remove URLs
def remove_url(text):
  pattern = re.compile(r'https?://\S+|www\.\S+')
  return pattern.sub(r'', text)

text = "Visit https://huggingface.co or http://example.com or www.google.com"
print(remove_url(text))
```

```python
# 4. punctuation
import string, time
excluded = string.punctuation

# slow
# def remove_punctuation(text):
#   for char in excluded:
#     text = text.replace(char, "")
#   return text

def remove_punctuation(text):
  return text.translate(str.maketrans('', '', string.punctuation))

text = "Hello! I love NLP, Python & AI."
cleaned_text = remove_punctuation(text)
print(cleaned_text)
```

```python
# 5. spelling correction
from textblob import TextBlob

def correct_spelling(text):
    blob = TextBlob(text)
    return str(blob.correct())

# Example usage
text = "I realy lovee natual langauge proccessing!"
corrected_text = correct_spelling(text)
print(corrected_text)

```
